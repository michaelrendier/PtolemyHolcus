````python
# R4 visualised: persistence is an octonion at every scale; the void grows.
import matplotlib.pyplot as plt
import numpy as np
plt.rcParams.update({'figure.dpi': 120, 'font.size': 10})

eng = GenerationalLineageEngine()
dims = [8, 16, 32, 64]
collapse, persist, amplify = [], [], []
for d in dims:
    sp = eng.gains(d, 1, d // 2 + 2)
    collapse.append(sp.get(0.0, 0))
    persist.append(sp.get(1.0, 0))
    amplify.append(sp.get(round(np.sqrt(2), 4), 0))

x = np.arange(len(dims))
fig, ax = plt.subplots(figsize=(7, 4))
ax.bar(x, collapse, label='gain 0  (collapse — the void)', color='#3b5bdb')
ax.bar(x, persist, bottom=collapse, label='gain 1  (PERSIST — the octonion core)', color='#e8a838')
ax.bar(x, amplify, bottom=[c+p for c, p in zip(collapse, persist)], label='gain √2  (amplify)', color='#c92a2a')
ax.set_xticks(x); ax.set_xticklabels([f'CD {d}' for d in dims])
ax.set_ylabel('dimensions'); ax.set_title('persist ≡ 8 (an octonion) at every scale; void = (d−8)/2')
ax.legend(loc='upper left', fontsize=8)
for i, d in enumerate(dims):
    ax.text(i, d + 1.5, f'8/{d}={8/d:.3f}', ha='center', fontsize=8)
plt.tight_layout(); plt.show()
print('persist:', persist, ' — constant 8. fraction 8/d → 0, the d*_RG flow.')
````
